```python
import numpy as np
from rfsoc_awg import *
from rfsoc_dashboard import AWGDashboard
```

```python
from pynq import Overlay, PL
import xrfclk
from adc_scope import AdcScope

PL.reset()
xrfclk.set_ref_clks(lmk_freq=245.76, lmx_freq=491.52)
ol = Overlay("awg_adc_v5.bit")

awg = AWG(window=1 * us, bitfile=ol, verbose=False)   # window = length of one AWG slot
_ = awg.connect()
scope = AdcScope(ol)
```

```python
dash = AWGDashboard(awg, scope=scope, adc_vfs=0.1/0.17) # Measurement VFS to redo
dash.ui

# NOTE may want to do frequency=0 for the cal to work.
```
